```python
# ==========================
# LOAD DATASET
# ==========================

train = pd.read_csv("Train_data.csv")
test = pd.read_csv("Test_data.csv")

print(train.shape)
print(test.shape)

train.head()
```

```python
# Dataset Information

train.info()
```

```python
# Missing Values

train.isnull().sum()
```

```python
# Duplicate Values

print(train.duplicated().sum())
```

```python
# Encode Target Column

train['class'] = train['class'].map({
    'normal':0,
    'anomaly':1
})
```

```python
# Encode Categorical Columns

encoder = LabelEncoder()

categorical_columns = ['protocol_type','service','flag']

for col in categorical_columns:
    train[col] = encoder.fit_transform(train[col])
    test[col] = encoder.fit_transform(test[col])
```

```python
# ==========================================================
# FEATURES AND TARGET
# ==========================================================

X = train.drop("class", axis=1)
y = train["class"]

# ==========================================================
# FEATURE SCALING
# ==========================================================

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X = scaler.fit_transform(X)

# Remove Prediction column if it already exists
if "Prediction" in test.columns:
    test = test.drop(columns=["Prediction"])

test_scaled = scaler.transform(test)

# ==========================================================
# TRAIN TEST SPLIT
# ==========================================================

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# ==========================================================
# IMPORT LIBRARIES
# ==========================================================

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

import matplotlib.pyplot as plt
import seaborn as sns

# ==========================================================
# LOGISTIC REGRESSION
# ==========================================================

lr = LogisticRegression(max_iter=1000)

lr.fit(X_train, y_train)

pred_lr = lr.predict(X_test)

lr_accuracy = accuracy_score(y_test, pred_lr)

print("\n==============================")
print("LOGISTIC REGRESSION")
print("==============================")

print("Accuracy :", lr_accuracy)

print(classification_report(y_test, pred_lr))

# ==========================================================
# DECISION TREE
# ==========================================================

dt = DecisionTreeClassifier(random_state=42)

dt.fit(X_train, y_train)

pred_dt = dt.predict(X_test)

dt_accuracy = accuracy_score(y_test, pred_dt)

print("\n==============================")
print("DECISION TREE")
print("==============================")

print("Accuracy :", dt_accuracy)

print(classification_report(y_test, pred_dt))

# ==========================================================
# RANDOM FOREST
# ==========================================================

rf = RandomForestClassifier(
    n_estimators=200,
    random_state=42
)

rf.fit(X_train, y_train)

pred_rf = rf.predict(X_test)

rf_accuracy = accuracy_score(y_test, pred_rf)

print("\n==============================")
print("RANDOM FOREST")
print("==============================")

print("Accuracy :", rf_accuracy)

print(classification_report(y_test, pred_rf))

# ==========================================================
# CONFUSION MATRIX
# ==========================================================

cm = confusion_matrix(y_test, pred_rf)

plt.figure(figsize=(6,5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues"
)

plt.title("Random Forest Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.show()

# ==========================================================
# FEATURE IMPORTANCE
# ==========================================================

importance = pd.Series(
    rf.feature_importances_,
    index=train.drop("class", axis=1).columns
)

importance = importance.sort_values(ascending=False)

plt.figure(figsize=(12,6))

importance.head(15).plot(kind="bar")

plt.title("Top 15 Important Features")

plt.show()

# ==========================================================
# MODEL COMPARISON
# ==========================================================

models = [
    "Logistic Regression",
    "Decision Tree",
    "Random Forest"
]

accuracy = [
    lr_accuracy,
    dt_accuracy,
    rf_accuracy
]

plt.figure(figsize=(8,5))

plt.bar(models, accuracy)

plt.ylabel("Accuracy")

plt.title("Accuracy Comparison")

plt.show()

print("\nAccuracy Comparison")

for model, acc in zip(models, accuracy):
    print(model, ":", round(acc*100,2), "%")

# ==========================================================
# SAVE MODEL
# ==========================================================

import joblib

joblib.dump(rf, "Threat_Detection_Model.pkl")

joblib.dump(scaler, "Scaler.pkl")

print("\nModel Saved Successfully")

# ==========================================================
# PREDICT TEST DATA
# ==========================================================

prediction = rf.predict(test_scaled)

prediction = pd.Series(prediction).replace({
    0: "Normal",
    1: "Attack"
})

test_result = test.copy()

test_result["Prediction"] = prediction

print(test_result.head())

# ==========================================================
# SAVE PREDICTION FILE
# ==========================================================

test_result.to_csv(
    "Threat_Predictions.csv",
    index=False
)

print("\nPrediction File Saved Successfully")

print("\n===================================")
print("AI-Assisted Threat Detection Completed")
print("Best Model : Random Forest")
print("Accuracy :", round(rf_accuracy*100,2), "%")
print("===================================")
```
