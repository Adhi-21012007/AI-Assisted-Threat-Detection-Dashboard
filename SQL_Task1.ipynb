Create a single table with 5–6 meaningful columns

```python
%%sql

CREATE TABLE students (
    student_id INTEGER PRIMARY KEY,
    name TEXT,
    age INTEGER,
    department TEXT,
    marks INTEGER
);
```

Display all records (SELECT *)

```python
%%sql

SELECT * FROM students;
```

Select specific columns

```python
%%sql

SELECT name, marks
FROM students;
```

Filter using WHERE

```python
%%sql

SELECT *
FROM students
WHERE marks > 80;
```

Sort using ORDER BY

```python
%%sql

SELECT *
FROM students
ORDER BY marks DESC;
```

Count records using COUNT()

```python
%%sql

SELECT COUNT(*) AS total_students
FROM students;
```

Find average using AVG()

```python
%%sql

SELECT AVG(marks) AS average_marks
FROM students;
```

Find maximum value using MAX()

```python
%%sql

SELECT MAX(marks) AS highest_marks
FROM students;
```

Filter using another condition (WHERE)

```python
%%sql

SELECT *
FROM students
WHERE department = 'CSE';
```

Group data using GROUP BY

```python
%%sql

SELECT department,
COUNT(*) AS total_students
FROM students
GROUP BY department;
```

Use CASE WHEN to classify records

```python
%%sql

SELECT
name,
marks,
CASE
    WHEN marks >= 90 THEN 'A'
    WHEN marks >= 80 THEN 'B'
    WHEN marks >= 70 THEN 'C'
    ELSE 'D'
END AS grade
FROM students;
```

The query should update an existing record in the students table by modifying the marks column for a specific student based on their student_id.

```python
%%sql

UPDATE students
SET marks = 90
WHERE student_id = 107;
```
