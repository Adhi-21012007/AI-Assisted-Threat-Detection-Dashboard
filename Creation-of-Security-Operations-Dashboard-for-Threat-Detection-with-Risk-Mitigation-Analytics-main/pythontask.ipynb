Input and Output

Prompt

Objective: Create a Python program that accepts user input for multiple fields, processes the input appropriately, and displays a formatted output using string formatting techniques.

```python
# ==============================================================
# Topic: Input and Output in Python
# Objective:
# Create a Python program that accepts user input for multiple
# fields, processes the input appropriately, and displays a
# formatted output using string formatting techniques.
# ==============================================================

print("=" * 70)
print("           PYTHON INPUT AND OUTPUT DEMONSTRATION")
print("=" * 70)

# ----------------------------------------------------
# Accepting User Input
# ----------------------------------------------------

print("\nPlease enter the following details:\n")

student_name = input("Enter Student Name        : ")
student_age = int(input("Enter Age                 : "))
college_name = input("Enter College Name        : ")
course = input("Enter Course              : ")
python_marks = float(input("Enter Python Marks        : "))
sql_marks = float(input("Enter SQL Marks           : "))

# ----------------------------------------------------
# Processing the Input
# ----------------------------------------------------

average_marks = (python_marks + sql_marks) / 2

if average_marks >= 90:
    grade = "A+"
elif average_marks >= 80:
    grade = "A"
elif average_marks >= 70:
    grade = "B"
elif average_marks >= 60:
    grade = "C"
else:
    grade = "D"

# ----------------------------------------------------
# Displaying Formatted Output
# ----------------------------------------------------

print("\n" + "=" * 70)
print("                 STUDENT ACADEMIC REPORT")
print("=" * 70)

print(f"Student Name     : {student_name}")
print(f"Age              : {student_age}")
print(f"College Name     : {college_name}")
print(f"Course           : {course}")
print(f"Python Marks     : {python_marks}")
print(f"SQL Marks        : {sql_marks}")
print(f"Average Marks    : {average_marks:.2f}")
print(f"Final Grade      : {grade}")

print("=" * 70)

# ----------------------------------------------------
# Displaying Output Using Different Formatting Techniques
# ----------------------------------------------------

print("\nFormatted Output Using format() Method")
print("Student: {} | Average: {:.2f} | Grade: {}".format(student_name, average_marks, grade))

print("\nFormatted Output Using f-Strings")
print(f"Congratulations {student_name}! Your average score is {average_marks:.2f} and your grade is '{grade}'.")

print("\nProgram executed successfully.")
```

- Understood the importance of converting user input into appropriate data types using `int()` and `float()`.
- Processed multiple user inputs to perform calculations and generate meaningful results.
- Applied conditional statements to determine grades based on calculated average marks.
- Explored different output formatting techniques, including **f-strings** and the `format()` method.
- Improved the readability of program output through proper formatting and structured display.
- Recognized the significance of input processing and formatted output in developing interactive Python applications.

```python
# ==============================================================
# Topic: Python Sets
# Objective:
# Create a Python program demonstrating set creation and
# manipulation using union, intersection, difference,
# symmetric difference, subset verification, superset
# verification, and duplicate elimination.
# ==============================================================

print("=" * 80)
print("                  PYTHON SET OPERATIONS")
print("=" * 80)

# ----------------------------------------------------
# Set Creation
# ----------------------------------------------------

backend_skills = {"Python", "SQL", "Django", "Git"}
frontend_skills = {"HTML", "CSS", "JavaScript", "Git"}

print("\n1. Set Creation")
print("-" * 80)
print("Backend Skills :", backend_skills)
print("Frontend Skills:", frontend_skills)

# ====================================================
# Duplicate Elimination
# ====================================================

print("\n2. Duplicate Elimination")
print("-" * 80)

programming_languages = [
    "Python", "Java", "Python", "SQL",
    "Java", "C++", "Python"
]

print("Original List          :", programming_languages)

unique_languages = set(programming_languages)

print("After Removing Duplicates:", unique_languages)

# ====================================================
# Union
# ====================================================

print("\n3. Union Operation")
print("-" * 80)

all_skills = backend_skills.union(frontend_skills)

print("Union of Sets :", all_skills)

# ====================================================
# Intersection
# ====================================================

print("\n4. Intersection Operation")
print("-" * 80)

common_skills = backend_skills.intersection(frontend_skills)

print("Common Skills :", common_skills)

# ====================================================
# Difference
# ====================================================

print("\n5. Difference Operation")
print("-" * 80)

backend_only = backend_skills.difference(frontend_skills)

print("Backend Only Skills :", backend_only)

# ====================================================
# Symmetric Difference
# ====================================================

print("\n6. Symmetric Difference")
print("-" * 80)

unique_skills = backend_skills.symmetric_difference(frontend_skills)

print("Unique Skills :", unique_skills)

# ====================================================
# Subset Verification
# ====================================================

print("\n7. Subset Verification")
print("-" * 80)

basic_python = {"Python", "Git"}

print("Basic Skills :", basic_python)
print("Is Subset?   :", basic_python.issubset(backend_skills))

# ====================================================
# Superset Verification
# ====================================================

print("\n8. Superset Verification")
print("-" * 80)

print("Is Superset? :", backend_skills.issuperset(basic_python))

# ====================================================
# Common Set Methods
# ====================================================

print("\n9. Common Set Methods")
print("-" * 80)

technologies = {"Python", "SQL"}

print("Original Set :", technologies)

technologies.add("Pandas")
print("After add() :", technologies)

technologies.update(["NumPy", "Machine Learning"])
print("After update() :", technologies)

technologies.remove("SQL")
print("After remove() :", technologies)

removed_item = technologies.pop()
print(f"Removed Item : {removed_item}")
print("After pop()  :", technologies)

# ====================================================
# Traversing a Set
# ====================================================

print("\n10. Traversing a Set")
print("-" * 80)

for skill in technologies:
    print(skill)

# ====================================================
# Membership Operators
# ====================================================

print("\n11. Membership Test")
print("-" * 80)

print("Python in technologies :", "Python" in technologies)
print("Java in technologies   :", "Java" in technologies)

# ====================================================
# Summary
# ====================================================

print("\n" + "=" * 80)
print("All Set Operations Executed Successfully")
print("=" * 80)
```

```python
# ==============================================================
# Topic: Python Loops
# Objective:
# Demonstrate iterative programming using both for and while
# loops by generating sequences, traversing collections,
# and implementing loop control statements such as break,
# continue, and pass.
# ==============================================================

print("=" * 80)
print("                  PYTHON LOOPS DEMONSTRATION")
print("=" * 80)

# ----------------------------------------------------
# Example 1: for Loop
# ----------------------------------------------------

print("\n1. for Loop")
print("-" * 80)

for number in range(1, 6):
    print(f"Iteration : {number}")

# ====================================================
# Example 2: while Loop
# ====================================================

print("\n2. while Loop")
print("-" * 80)

count = 1

while count <= 5:
    print(f"Count : {count}")
    count += 1

# ====================================================
# Example 3: Traversing a List
# ====================================================

print("\n3. Traversing a List")
print("-" * 80)

skills = ["Python", "SQL", "NumPy", "Pandas"]

for skill in skills:
    print(skill)

# ====================================================
# Example 4: enumerate()
# ====================================================

print("\n4. enumerate()")
print("-" * 80)

for index, skill in enumerate(skills, start=1):
    print(f"{index}. {skill}")

# ====================================================
# Example 5: break Statement
# ====================================================

print("\n5. break Statement")
print("-" * 80)

for number in range(1, 11):

    if number == 6:
        print("Loop terminated using break.")
        break

    print(number)

# ====================================================
# Example 6: continue Statement
# ====================================================

print("\n6. continue Statement")
print("-" * 80)

for number in range(1, 11):

    if number % 2 == 0:
        continue

    print(number)

# ====================================================
# Example 7: pass Statement
# ====================================================

print("\n7. pass Statement")
print("-" * 80)

for number in range(1, 6):

    if number == 3:
        pass

    print(number)

# ====================================================
# Example 8: Nested Loops
# ====================================================

print("\n8. Nested Loops")
print("-" * 80)

for row in range(1, 4):
    for column in range(1, 4):
        print(f"({row},{column})", end=" ")
    print()

# ====================================================
# Example 9: Multiplication Table
# ====================================================

print("\n9. Multiplication Table")
print("-" * 80)

table = 5

for i in range(1, 11):
    print(f"{table} x {i} = {table * i}")

# ====================================================
# Example 10: Sum of Numbers
# ====================================================

print("\n10. Sum of Numbers")
print("-" * 80)

total = 0

for number in range(1, 11):
    total += number

print("Sum =", total)

# ====================================================
# Example 11: Factorial
# ====================================================

print("\n11. Factorial")
print("-" * 80)

number = 5
factorial = 1

for i in range(1, number + 1):
    factorial *= i

print(f"Factorial of {number} = {factorial}")

# ====================================================
# Example 12: Pattern Printing
# ====================================================

print("\n12. Pattern Printing")
print("-" * 80)

rows = 5

for i in range(1, rows + 1):
    print("*" * i)

# ====================================================
# Example 13: Countdown using while
# ====================================================

print("\n13. Countdown")
print("-" * 80)

count = 5

while count > 0:
    print(count)
    count -= 1

print("Liftoff 🚀")

# ====================================================
# Example 14: Searching an Element
# ====================================================

print("\n14. Searching in List")
print("-" * 80)

languages = ["Python", "Java", "SQL", "C++"]

search = "SQL"

for language in languages:

    if language == search:
        print(f"{search} Found")
        break

# ====================================================
# Example 15: Real-World Example
# ====================================================

print("\n15. Student Marks")
print("-" * 80)

marks = [91, 84, 78, 96, 88]

average = sum(marks) / len(marks)

print("Marks :", marks)

for mark in marks:
    print(mark)

print(f"Average : {average:.2f}")

# ====================================================
# Summary
# ====================================================

print("\n" + "=" * 80)
print("All Loop Examples Executed Successfully")
print("=" * 80)
```

```python
# ==============================================================
# Topic: Python Lambda Functions
# Objective:
# Implement anonymous functions using Python's lambda
# expression and demonstrate their application with
# built-in functions such as map(), filter(), sorted(),
# and other practical examples.
# ==============================================================

from functools import reduce

print("=" * 80)
print("              PYTHON LAMBDA FUNCTIONS")
print("=" * 80)

# ----------------------------------------------------
# Example 1: Basic Lambda Function
# ----------------------------------------------------

print("\n1. Basic Lambda Function")
print("-" * 80)

square = lambda number: number ** 2

print("Square of 8 :", square(8))
print("Square of 15:", square(15))

# ====================================================
# Example 2: Lambda with Multiple Arguments
# ====================================================

print("\n2. Lambda with Multiple Arguments")
print("-" * 80)

addition = lambda a, b: a + b

print("25 + 15 =", addition(25, 15))

# ====================================================
# Example 3: map()
# ====================================================

print("\n3. map() Function")
print("-" * 80)

numbers = [2, 4, 6, 8, 10]

squares = list(map(lambda number: number ** 2, numbers))

print("Original Numbers :", numbers)
print("Squared Numbers  :", squares)

# ====================================================
# Example 4: filter()
# ====================================================

print("\n4. filter() Function")
print("-" * 80)

marks = [45, 92, 78, 34, 88, 59, 99]

passed_students = list(filter(lambda mark: mark >= 60, marks))

print("All Marks      :", marks)
print("Passed Students:", passed_students)

# ====================================================
# Example 5: sorted()
# ====================================================

print("\n5. sorted() with Lambda")
print("-" * 80)

students = [
    ("Rahul", 85),
    ("Anjali", 95),
    ("Kiran", 78),
    ("Srilekha", 98)
]

sorted_students = sorted(
    students,
    key=lambda student: student[1],
    reverse=True
)

print("Students Sorted by Marks")

for name, marks in sorted_students:
    print(f"{name:<12} {marks}")

# ====================================================
# Example 6: reduce()
# ====================================================

print("\n6. reduce() Function")
print("-" * 80)

numbers = [10, 20, 30, 40]

total = reduce(lambda x, y: x + y, numbers)

print("Numbers :", numbers)
print("Sum     :", total)

# ====================================================
# Example 7: Sorting Dictionary
# ====================================================

print("\n7. Sorting Dictionary")
print("-" * 80)

employees = {
    "Rahul": 65000,
    "Priya": 85000,
    "Anjali": 72000,
    "Srilekha": 95000
}

sorted_salary = sorted(
    employees.items(),
    key=lambda employee: employee[1],
    reverse=True
)

for name, salary in sorted_salary:
    print(f"{name:<12} ₹{salary}")

# ====================================================
# Example 8: Maximum Value
# ====================================================

print("\n8. Maximum Value using Lambda")
print("-" * 80)

numbers = [55, 18, 92, 34, 77]

maximum = max(numbers, key=lambda number: number)

print("Numbers :", numbers)
print("Maximum :", maximum)

# ====================================================
# Example 9: String Sorting
# ====================================================

print("\n9. Sorting Strings by Length")
print("-" * 80)

languages = [
    "Python",
    "Java",
    "JavaScript",
    "C",
    "SQL"
]

sorted_languages = sorted(
    languages,
    key=lambda language: len(language)
)

print(sorted_languages)

# ====================================================
# Example 10: Real-World Example
# ====================================================

print("\n10. Student Percentage")
print("-" * 80)

students = [
    {"Name": "Rahul", "Marks": 91},
    {"Name": "Priya", "Marks": 84},
    {"Name": "Srilekha", "Marks": 98},
    {"Name": "Anjali", "Marks": 88}
]

top_students = list(
    filter(
        lambda student: student["Marks"] >= 90,
        students
    )
)

print("Students Scoring 90 or Above")

for student in top_students:
    print(student)

# ====================================================
# Example 11: Lambda with Conditional Expression
# ====================================================

print("\n11. Conditional Lambda")
print("-" * 80)

even_or_odd = lambda number: "Even" if number % 2 == 0 else "Odd"

print("12 is", even_or_odd(12))
print("17 is", even_or_odd(17))

# ====================================================
# Example 12: Sorting Objects by Multiple Keys
# ====================================================

print("\n12. Multiple-Key Sorting")
print("-" * 80)

employees = [
    ("Rahul", "AI", 70000),
    ("Priya", "Python", 85000),
    ("Anjali", "AI", 82000),
    ("Kiran", "Python", 75000)
]

sorted_employees = sorted(
    employees,
    key=lambda employee: (employee[1], employee[2])
)

for employee in sorted_employees:
    print(employee)

# ====================================================
# Summary
# ====================================================

print("\n" + "=" * 80)
print("All Lambda Function Examples Executed Successfully")
print("=" * 80)
```

```python
# ==============================================================
# Topic: Object-Oriented Programming (OOP)
# Part 1
# Demonstrates:
# ✔ Classes & Objects
# ✔ Constructors
# ✔ Instance Variables
# ✔ Class Variables
# ✔ Instance Methods
# ✔ Encapsulation
# ✔ Single Inheritance
# ==============================================================

from abc import ABC, abstractmethod

print("=" * 80)
print("         OBJECT ORIENTED PROGRAMMING (OOP)")
print("=" * 80)

# ==========================================================
# Example 1 : Classes and Objects
# ==========================================================

print("\n1. Classes and Objects")
print("-" * 80)


class Student:

    # Class Variable
    college = "Infosys Springboard Academy"

    # Constructor
    def __init__(self, student_id, name, course, cgpa):
        self.student_id = student_id
        self.name = name
        self.course = course
        self.cgpa = cgpa

    # Instance Method
    def display_details(self):
        print(f"Student ID : {self.student_id}")
        print(f"Name       : {self.name}")
        print(f"Course     : {self.course}")
        print(f"CGPA       : {self.cgpa}")
        print(f"College    : {Student.college}")


student1 = Student(101, "Srilekha", "Python", 9.45)

student1.display_details()

# ==========================================================
# Example 2 : Updating Instance Variables
# ==========================================================

print("\n2. Updating Instance Variables")
print("-" * 80)

student1.cgpa = 9.60

student1.display_details()

# ==========================================================
# Example 3 : Class Variables
# ==========================================================

print("\n3. Class Variables")
print("-" * 80)

student2 = Student(102, "Rahul", "AI", 8.95)

print("College of Student 1 :", student1.college)
print("College of Student 2 :", student2.college)

# ==========================================================
# Example 4 : Encapsulation
# ==========================================================

print("\n4. Encapsulation")
print("-" * 80)


class BankAccount:

    def __init__(self, account_holder, balance):

        self.account_holder = account_holder

        # Private Variable
        self.__balance = balance

    def deposit(self, amount):

        self.__balance += amount

        print(f"₹{amount} deposited successfully.")

    def withdraw(self, amount):

        if amount <= self.__balance:

            self.__balance -= amount

            print(f"₹{amount} withdrawn successfully.")

        else:

            print("Insufficient Balance.")

    def get_balance(self):

        return self.__balance


account = BankAccount("Srilekha", 5000)

account.deposit(2500)

account.withdraw(1500)

print("Available Balance :", account.get_balance())

# ==========================================================
# Example 5 : Single Inheritance
# ==========================================================

print("\n5. Single Inheritance")
print("-" * 80)


class Person:

    def __init__(self, name):

        self.name = name

    def display(self):

        print("Person Name :", self.name)


class Employee(Person):

    def __init__(self, name, employee_id, department):

        super().__init__(name)

        self.employee_id = employee_id
        self.department = department

    def employee_details(self):

        print("Employee ID :", self.employee_id)
        print("Department  :", self.department)


employee = Employee(
    "Anjali",
    "EMP101",
    "Cyber Security"
)

employee.display()

employee.employee_details()

# ==========================================================
# Example 6 : Method Demonstration
# ==========================================================

print("\n6. Instance Methods")
print("-" * 80)


class Rectangle:

    def __init__(self, length, width):

        self.length = length
        self.width = width

    def area(self):

        return self.length * self.width

    def perimeter(self):

        return 2 * (self.length + self.width)


rectangle = Rectangle(12, 6)

print("Area :", rectangle.area())

print("Perimeter :", rectangle.perimeter())

print("\n" + "=" * 80)
print("End of OOP Part 1")
print("=" * 80)
```

```python
# ==============================================================
# Topic: Object-Oriented Programming (OOP)
# Part 2
# Demonstrates:
# ✔ Multilevel Inheritance
# ✔ Polymorphism (Method Overriding)
# ✔ Abstraction
# ✔ Real-World Banking System
# ==============================================================

from abc import ABC, abstractmethod

print("=" * 80)
print("        ADVANCED OBJECT-ORIENTED PROGRAMMING (OOP)")
print("=" * 80)

# ==========================================================
# Example 7 : Multilevel Inheritance
# ==========================================================

print("\n7. Multilevel Inheritance")
print("-" * 80)


class Person:

    def display_person(self):
        print("Person Information")


class Employee(Person):

    def display_employee(self):
        print("Employee Information")


class Manager(Employee):

    def display_manager(self):
        print("Manager Information")


manager = Manager()

manager.display_person()
manager.display_employee()
manager.display_manager()

# ==========================================================
# Example 8 : Polymorphism (Method Overriding)
# ==========================================================

print("\n8. Polymorphism (Method Overriding)")
print("-" * 80)


class Animal:

    def sound(self):
        print("Animals produce different sounds.")


class Dog(Animal):

    def sound(self):
        print("Dog says: Bark")


class Cat(Animal):

    def sound(self):
        print("Cat says: Meow")


animals = [Dog(), Cat()]

for animal in animals:
    animal.sound()

# ==========================================================
# Example 9 : Abstraction
# ==========================================================

print("\n9. Abstraction")
print("-" * 80)


class Vehicle(ABC):

    @abstractmethod
    def start_engine(self):
        pass


class Car(Vehicle):

    def start_engine(self):
        print("Car engine started successfully.")


class Bike(Vehicle):

    def start_engine(self):
        print("Bike engine started successfully.")


car = Car()
bike = Bike()

car.start_engine()
bike.start_engine()

# ==========================================================
# Example 10 : Real-World Banking System
# ==========================================================

print("\n10. Real-World Banking System")
print("-" * 80)


class BankAccount:

    bank_name = "ABC National Bank"

    def __init__(self, account_number, account_holder, balance):

        self.account_number = account_number
        self.account_holder = account_holder
        self.__balance = balance

    def deposit(self, amount):

        if amount > 0:
            self.__balance += amount
            print(f"₹{amount} deposited successfully.")

    def withdraw(self, amount):

        if amount <= self.__balance:
            self.__balance -= amount
            print(f"₹{amount} withdrawn successfully.")
        else:
            print("Transaction Failed: Insufficient Balance.")

    def display_account(self):

        print(f"Bank Name      : {BankAccount.bank_name}")
        print(f"Account Number : {self.account_number}")
        print(f"Account Holder : {self.account_holder}")
        print(f"Available Balance : ₹{self.__balance}")


account1 = BankAccount(
    "SBI1001",
    "Srilekha",
    25000
)

account1.display_account()

print()

account1.deposit(5000)

account1.withdraw(8000)

print()

account1.display_account()

# ==========================================================
# Example 11 : Polymorphism Using Different Shapes
# ==========================================================

print("\n11. Polymorphism Example")
print("-" * 80)


class Shape:

    def area(self):
        pass


class Circle(Shape):

    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius * self.radius


class Rectangle(Shape):

    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width


shapes = [
    Circle(5),
    Rectangle(10, 6)
]

for shape in shapes:
    print("Area =", shape.area())

# ==========================================================
# Example 12 : Class Method
# ==========================================================

print("\n12. Class Method")
print("-" * 80)


class Company:

    company_name = "Infosys"

    @classmethod
    def display_company(cls):
        print("Company Name :", cls.company_name)


Company.display_company()

# ==========================================================
# Example 13 : Static Method
# ==========================================================

print("\n13. Static Method")
print("-" * 80)


class Calculator:

    @staticmethod
    def add(a, b):
        return a + b


print("Addition =", Calculator.add(15, 20))

# ==========================================================
# Summary
# ==========================================================

print("\n" + "=" * 80)
print("All Advanced OOP Concepts Executed Successfully")
print("=" * 80)
```

```python
# ==============================================================
# Topic: Python Exception Handling
# Objective:
# Create a robust Python program capable of handling runtime
# exceptions using try, except, else, finally, and user-defined
# exception handling mechanisms.
# ==============================================================

print("=" * 80)
print("              PYTHON EXCEPTION HANDLING")
print("=" * 80)

# ==========================================================
# Example 1 : Handling ZeroDivisionError
# ==========================================================

print("\n1. Handling ZeroDivisionError")
print("-" * 80)

try:
    numerator = 100
    denominator = 5

    result = numerator / denominator

    print("Result :", result)

except ZeroDivisionError:
    print("Error: Division by zero is not allowed.")

# ==========================================================
# Example 2 : Handling ValueError
# ==========================================================

print("\n2. Handling ValueError")
print("-" * 80)

try:
    age = int("Twenty")

    print(age)

except ValueError:
    print("Error: Invalid integer value.")

# ==========================================================
# Example 3 : Handling IndexError
# ==========================================================

print("\n3. Handling IndexError")
print("-" * 80)

try:
    languages = ["Python", "Java", "SQL"]

    print(languages[5])

except IndexError:
    print("Error: List index is out of range.")

# ==========================================================
# Example 4 : Handling KeyError
# ==========================================================

print("\n4. Handling KeyError")
print("-" * 80)

try:
    student = {
        "Name": "Srilekha",
        "Course": "Python"
    }

    print(student["CGPA"])

except KeyError:
    print("Error: Requested key does not exist.")

# ==========================================================
# Example 5 : Handling Multiple Exceptions
# ==========================================================

print("\n5. Handling Multiple Exceptions")
print("-" * 80)

try:
    number = int("ABC")

    print(100 / number)

except ValueError:
    print("ValueError occurred.")

except ZeroDivisionError:
    print("ZeroDivisionError occurred.")

# ==========================================================
# Example 6 : else Block
# ==========================================================

print("\n6. else Block")
print("-" * 80)

try:
    marks = [92, 88, 95]

    print("Highest Marks :", max(marks))

except ValueError:
    print("List is empty.")

else:
    print("Program executed successfully.")

# ==========================================================
# Example 7 : finally Block
# ==========================================================

print("\n7. finally Block")
print("-" * 80)

try:
    file = open("sample.txt", "w")

    file.write("Python Exception Handling Example")

    print("Data written successfully.")

except Exception as error:
    print(error)

finally:
    file.close()

    print("File closed successfully.")

# ==========================================================
# Example 8 : Raising an Exception
# ==========================================================

print("\n8. Raising an Exception")
print("-" * 80)

try:
    age = 15

    if age < 18:
        raise ValueError("Age must be at least 18.")

    print("Eligible.")

except ValueError as error:
    print("Error:", error)

# ==========================================================
# Example 9 : User-Defined Exception
# ==========================================================

print("\n9. User-Defined Exception")
print("-" * 80)


class InsufficientBalanceError(Exception):
    """Custom Exception for insufficient account balance."""
    pass


class BankAccount:

    def __init__(self, balance):
        self.balance = balance

    def withdraw(self, amount):

        if amount > self.balance:
            raise InsufficientBalanceError(
                "Withdrawal amount exceeds available balance."
            )

        self.balance -= amount

        print(f"₹{amount} withdrawn successfully.")
        print(f"Remaining Balance : ₹{self.balance}")


account = BankAccount(5000)

try:
    account.withdraw(6000)

except InsufficientBalanceError as error:
    print("Transaction Failed:", error)

# ==========================================================
# Example 10 : Generic Exception
# ==========================================================

print("\n10. Generic Exception")
print("-" * 80)

try:
    numbers = [10, 20, 30]

    print(numbers[10])

except Exception as error:
    print("Unexpected Error:", error)

# ==========================================================
# Example 11 : Real-World Example
# ==========================================================

print("\n11. Student Result Processing")
print("-" * 80)

student_marks = {
    "Python": 95,
    "SQL": 88,
    "NumPy": 91
}

try:
    subject = "Python"

    print(f"{subject} Marks :", student_marks[subject])

except KeyError:
    print("Subject not found.")

else:
    print("Marks retrieved successfully.")

finally:
    print("Result processing completed.")

# ==========================================================
# Summary
# ==========================================================

print("\n" + "=" * 80)
print("All Exception Handling Examples Executed Successfully")
print("=" * 80)
```

```python
# ==============================================================
# Topic: Python File Handling
# Objective:
# Develop a Python application that performs file operations
# including reading, writing, appending, and managing files
# efficiently while implementing proper exception handling.
# ==============================================================

import os

print("=" * 80)
print("                  PYTHON FILE HANDLING")
print("=" * 80)

file_name = "student_data.txt"

# ==========================================================
# Example 1 : Writing to a File
# ==========================================================

print("\n1. Writing to a File")
print("-" * 80)

try:
    with open(file_name, "w") as file:
        file.write("Student Report\n")
        file.write("=" * 30 + "\n")
        file.write("Name   : Srilekha\n")
        file.write("Course : Python Programming\n")
        file.write("Marks  : 95\n")

    print("Data written successfully.")

except Exception as error:
    print("Error:", error)

# ==========================================================
# Example 2 : Reading the Entire File
# ==========================================================

print("\n2. Reading Entire File")
print("-" * 80)

try:
    with open(file_name, "r") as file:
        content = file.read()

    print(content)

except FileNotFoundError:
    print("File not found.")

# ==========================================================
# Example 3 : Reading Line by Line
# ==========================================================

print("\n3. Reading Line by Line")
print("-" * 80)

try:
    with open(file_name, "r") as file:
        for line in file:
            print(line.strip())

except FileNotFoundError:
    print("File not found.")

# ==========================================================
# Example 4 : Appending Data
# ==========================================================

print("\n4. Appending Data")
print("-" * 80)

try:
    with open(file_name, "a") as file:
        file.write("Grade  : A+\n")
        file.write("Status : Passed\n")

    print("Data appended successfully.")

except Exception as error:
    print(error)

# ==========================================================
# Example 5 : Reading Using readlines()
# ==========================================================

print("\n5. Reading Using readlines()")
print("-" * 80)

try:
    with open(file_name, "r") as file:
        lines = file.readlines()

    for line_number, line in enumerate(lines, start=1):
        print(f"{line_number}. {line.strip()}")

except FileNotFoundError:
    print("File not found.")

# ==========================================================
# Example 6 : File Information
# ==========================================================

print("\n6. File Information")
print("-" * 80)

if os.path.exists(file_name):
    print("File Exists :", file_name)
    print("File Size   :", os.path.getsize(file_name), "bytes")
else:
    print("File does not exist.")

# ==========================================================
# Example 7 : Copying File Content
# ==========================================================

print("\n7. Copying File Content")
print("-" * 80)

copy_file = "student_data_copy.txt"

try:
    with open(file_name, "r") as source:
        data = source.read()

    with open(copy_file, "w") as destination:
        destination.write(data)

    print("File copied successfully.")

except Exception as error:
    print(error)

# ==========================================================
# Example 8 : Counting Lines, Words and Characters
# ==========================================================

print("\n8. File Statistics")
print("-" * 80)

try:
    with open(file_name, "r") as file:
        content = file.read()

    lines = content.splitlines()
    words = content.split()
    characters = len(content)

    print("Total Lines      :", len(lines))
    print("Total Words      :", len(words))
    print("Total Characters :", characters)

except Exception as error:
    print(error)

# ==========================================================
# Example 9 : Searching a Word
# ==========================================================

print("\n9. Searching a Word")
print("-" * 80)

search_word = "Python"

try:
    with open(file_name, "r") as file:
        content = file.read()

    if search_word in content:
        print(f"'{search_word}' found in the file.")
    else:
        print(f"'{search_word}' not found.")

except Exception as error:
    print(error)

# ==========================================================
# Example 10 : Exception Handling
# ==========================================================

print("\n10. Exception Handling")
print("-" * 80)

try:
    with open("unknown_file.txt", "r") as file:
        print(file.read())

except FileNotFoundError:
    print("Error: The specified file does not exist.")

finally:
    print("File operation completed.")

# ==========================================================
# Example 11 : Removing a File
# ==========================================================

print("\n11. Removing a File")
print("-" * 80)

temporary_file = "temporary.txt"

with open(temporary_file, "w") as file:
    file.write("Temporary file for demonstration.")

if os.path.exists(temporary_file):
    os.remove(temporary_file)
    print("Temporary file deleted successfully.")
else:
    print("Temporary file not found.")

# ==========================================================
# Summary
# ==========================================================

print("\n" + "=" * 80)
print("All File Handling Operations Executed Successfully")
print("=" * 80)
```

```python
# ==============================================================
# Topic: NumPy
# Objective:
# Utilize the NumPy library to create multidimensional arrays
# and perform indexing, slicing, reshaping, aggregation,
# broadcasting, and mathematical computations efficiently.
# ==============================================================

import numpy as np

print("=" * 80)
print("                  NUMPY OPERATIONS")
print("=" * 80)

# ==========================================================
# Example 1 : Creating Arrays
# ==========================================================

print("\n1. Creating NumPy Arrays")
print("-" * 80)

array1 = np.array([10, 20, 30, 40, 50])
array2 = np.array([[1, 2, 3],
                   [4, 5, 6]])

print("One-Dimensional Array:")
print(array1)

print("\nTwo-Dimensional Array:")
print(array2)

# ==========================================================
# Example 2 : Array Attributes
# ==========================================================

print("\n2. Array Attributes")
print("-" * 80)

print("Dimensions :", array2.ndim)
print("Shape      :", array2.shape)
print("Size       :", array2.size)
print("Data Type  :", array2.dtype)

# ==========================================================
# Example 3 : Indexing
# ==========================================================

print("\n3. Array Indexing")
print("-" * 80)

print("First Element :", array1[0])
print("Last Element  :", array1[-1])
print("Element [1][2]:", array2[1][2])

# ==========================================================
# Example 4 : Slicing
# ==========================================================

print("\n4. Array Slicing")
print("-" * 80)

print("First Three Elements :", array1[:3])
print("Last Two Elements    :", array1[-2:])
print("Second Column        :", array2[:, 1])

# ==========================================================
# Example 5 : Reshaping
# ==========================================================

print("\n5. Reshaping Arrays")
print("-" * 80)

numbers = np.arange(1, 13)

matrix = numbers.reshape(3, 4)

print(matrix)

# ==========================================================
# Example 6 : Mathematical Operations
# ==========================================================

print("\n6. Mathematical Operations")
print("-" * 80)

a = np.array([5, 10, 15])
b = np.array([2, 4, 6])

print("Addition       :", a + b)
print("Subtraction    :", a - b)
print("Multiplication :", a * b)
print("Division       :", a / b)

# ==========================================================
# Example 7 : Aggregation Functions
# ==========================================================

print("\n7. Aggregation Functions")
print("-" * 80)

marks = np.array([91, 84, 76, 98, 89])

print("Sum     :", np.sum(marks))
print("Average :", np.mean(marks))
print("Maximum :", np.max(marks))
print("Minimum :", np.min(marks))
print("Median  :", np.median(marks))
print("Standard Deviation :", np.std(marks))

# ==========================================================
# Example 8 : Broadcasting
# ==========================================================

print("\n8. Broadcasting")
print("-" * 80)

numbers = np.array([1, 2, 3, 4, 5])

print("Original :", numbers)
print("After +10:", numbers + 10)

# ==========================================================
# Example 9 : Random Numbers
# ==========================================================

print("\n9. Random Arrays")
print("-" * 80)

np.random.seed(42)

random_numbers = np.random.randint(1, 100, size=(3, 3))

print(random_numbers)

# ==========================================================
# Example 10 : Matrix Operations
# ==========================================================

print("\n10. Matrix Operations")
print("-" * 80)

matrix1 = np.array([[1, 2],
                    [3, 4]])

matrix2 = np.array([[5, 6],
                    [7, 8]])

print("Matrix Addition:")
print(matrix1 + matrix2)

print("\nMatrix Multiplication:")
print(np.matmul(matrix1, matrix2))

print("\nTranspose:")
print(matrix1.T)

# ==========================================================
# Example 11 : Universal Functions
# ==========================================================

print("\n11. Universal Functions")
print("-" * 80)

angles = np.array([0, np.pi/2, np.pi])

print("Sine Values:")
print(np.sin(angles))

print("Square Root:")
print(np.sqrt(np.array([4, 16, 25])))

# ==========================================================
# Example 12 : Real-World Example
# ==========================================================

print("\n12. Student Marks Analysis")
print("-" * 80)

student_marks = np.array([
    [91, 85, 88],
    [76, 90, 82],
    [95, 94, 97],
    [81, 79, 84]
])

print("Student Marks:")
print(student_marks)

average_marks = np.mean(student_marks, axis=1)

print("\nAverage Marks:")
print(average_marks)

highest_score = np.max(student_marks)

print("\nHighest Score:", highest_score)

# ==========================================================
# Summary
# ==========================================================

print("\n" + "=" * 80)
print("All NumPy Operations Executed Successfully")
print("=" * 80)
```

Objective: Develop a comprehensive Python application integrating fundamental programming concepts, data structures, object-oriented programming, exception handling, file handling, and external libraries to solve a practical real-world problem.

```python
# ==============================================================
# Mini Project: Student Management System
#
# Objective:
# Develop a comprehensive Python application integrating
# Python fundamentals, OOP, Exception Handling,
# File Handling, and Pandas to manage student records.
# ==============================================================

import pandas as pd
import os

FILE_NAME = "students.csv"


class StudentManagementSystem:

    def __init__(self):

        if os.path.exists(FILE_NAME):
            self.students = pd.read_csv(FILE_NAME)
        else:
            self.students = pd.DataFrame(columns=[
                "Student_ID",
                "Name",
                "Department",
                "Age",
                "Marks",
                "Grade"
            ])

    # ----------------------------------------------------------
    # Save Data
    # ----------------------------------------------------------

    def save_data(self):
        self.students.to_csv(FILE_NAME, index=False)

    # ----------------------------------------------------------
    # Add Student
    # ----------------------------------------------------------

    def add_student(self):

        try:

            student_id = int(input("Enter Student ID : "))
            name = input("Enter Name : ")
            department = input("Enter Department : ")
            age = int(input("Enter Age : "))
            marks = float(input("Enter Marks : "))

            if marks >= 90:
                grade = "A+"
            elif marks >= 80:
                grade = "A"
            elif marks >= 70:
                grade = "B"
            elif marks >= 60:
                grade = "C"
            else:
                grade = "Fail"

            new_student = pd.DataFrame([{
                "Student_ID": student_id,
                "Name": name,
                "Department": department,
                "Age": age,
                "Marks": marks,
                "Grade": grade
            }])

            self.students = pd.concat(
                [self.students, new_student],
                ignore_index=True
            )

            self.save_data()

            print("\nStudent added successfully.")

        except ValueError:
            print("Invalid input.")

    # ----------------------------------------------------------
    # Display Students
    # ----------------------------------------------------------

    def display_students(self):

        if self.students.empty:
            print("\nNo records found.")
        else:
            print(self.students)

    # ----------------------------------------------------------
    # Search Student
    # ----------------------------------------------------------

    def search_student(self):

        student_id = int(input("Enter Student ID : "))

        student = self.students[
            self.students["Student_ID"] == student_id
        ]

        if student.empty:
            print("Student not found.")
        else:
            print(student)

    # ----------------------------------------------------------
    # Update Marks
    # ----------------------------------------------------------

    def update_marks(self):

        student_id = int(input("Enter Student ID : "))
        new_marks = float(input("Enter New Marks : "))

        index = self.students[
            self.students["Student_ID"] == student_id
        ].index

        if len(index) == 0:
            print("Student not found.")
            return

        self.students.loc[index, "Marks"] = new_marks

        if new_marks >= 90:
            grade = "A+"
        elif new_marks >= 80:
            grade = "A"
        elif new_marks >= 70:
            grade = "B"
        elif new_marks >= 60:
            grade = "C"
        else:
            grade = "Fail"

        self.students.loc[index, "Grade"] = grade

        self.save_data()

        print("Marks updated successfully.")

    # ----------------------------------------------------------
    # Delete Student
    # ----------------------------------------------------------

    def delete_student(self):

        student_id = int(input("Enter Student ID : "))

        self.students = self.students[
            self.students["Student_ID"] != student_id
        ]

        self.save_data()

        print("Student deleted successfully.")

    # ----------------------------------------------------------
    # Generate Report
    # ----------------------------------------------------------

    def generate_report(self):

        if self.students.empty:
            print("No data available.")
            return

        print("\n========== REPORT ==========")

        print("Total Students :", len(self.students))

        print("Average Marks :",
              round(self.students["Marks"].mean(), 2))

        print("Highest Marks :",
              self.students["Marks"].max())

        print("Lowest Marks :",
              self.students["Marks"].min())

        print("\nDepartment Wise Average")

        print(
            self.students.groupby("Department")["Marks"].mean()
        )


# ==============================================================
# Main Program
# ==============================================================

system = StudentManagementSystem()

while True:

    print("\n" + "=" * 60)
    print("      STUDENT MANAGEMENT SYSTEM")
    print("=" * 60)

    print("1. Add Student")
    print("2. Display Students")
    print("3. Search Student")
    print("4. Update Marks")
    print("5. Delete Student")
    print("6. Generate Report")
    print("7. Exit")

    choice = input("\nEnter your choice : ")

    if choice == "1":
        system.add_student()

    elif choice == "2":
        system.display_students()

    elif choice == "3":
        system.search_student()

    elif choice == "4":
        system.update_marks()

    elif choice == "5":
        system.delete_student()

    elif choice == "6":
        system.generate_report()

    elif choice == "7":
        print("\nThank you for using the Student Management System.")
        break

    else:
        print("Invalid choice.")
```
