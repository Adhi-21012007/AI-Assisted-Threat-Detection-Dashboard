CONNECT DATABASE

Prompt

Implement SQL operations using Python SQLite in Google Colab including table creation, insertion, filtering, aggregation, joins, and analysis.

```python
import sqlite3

conn = sqlite3.connect("student_db.sqlite")
cursor = conn.cursor()

print("Database connected successfully")
```

Key Learning
SQLite creates database file automatically
cursor is used to execute SQL commands
Connection is required for all operations

CREATE TABLE

Prompt

Objective: Create a structured table named students to store student details such as ID, name, department, marks, age, and city.



```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS students (
    student_id INTEGER PRIMARY KEY,
    name TEXT,
    department TEXT,
    age INTEGER,
    marks INTEGER,
    city TEXT
)
""")

print("Table created successfully")
```

Key Learning
Defines schema (structure of data)
IF NOT EXISTS avoids errors
Data types: INTEGER, TEXT

INSERT DATA

Prompt

Objective: Insert multiple student records into the students table using batch insertion.

```python
cursor.executemany("""
INSERT INTO students VALUES (?, ?, ?, ?, ?, ?)
""", [
    (101, 'Asha', 'AI', 20, 95, 'Hyderabad'),
    (102, 'Ravi', 'Cyber', 21, 88, 'Chennai'),
    (103, 'Meena', 'AI', 22, 91, 'Bangalore'),
    (104, 'Kiran', 'Data Science', 20, 84, 'Delhi'),
    (105, 'Priya', 'Cyber', 21, 89, 'Mumbai')
])

conn.commit()
print("Data inserted successfully")
```

Key Learning
executemany() inserts multiple rows
commit() saves changes permanently

SELECT ALL DATA

Prompt

Objective: Retrieve all records from the students table.

```python
cursor.execute("SELECT * FROM students")

rows = cursor.fetchall()
for row in rows:
    print(row)
```

Key Learning
SELECT retrieves data from table
fetchall() returns all rows

WHERE FILTER

Prompt

Objective: Filter students based on marks greater than 90.

```python
cursor.execute("""
SELECT * FROM students
WHERE marks > 90
""")

print(cursor.fetchall())
```

Key Learning
Filters rows based on condition
Used in almost all real queries

ORDER BY

Prompt

Objective: Sort students based on marks in descending order.

```python
cursor.execute("""
SELECT * FROM students
ORDER BY marks DESC
""")

print(cursor.fetchall())
```

Key Learning
Sorts data
DESC = highest to lowest

AND / OR

Prompt

Objective: Apply multiple conditions using logical operators.

```python
cursor.execute("""
SELECT * FROM students
WHERE department = 'AI' AND marks > 90
""")

print(cursor.fetchall())
```

Key Learning
AND → all conditions must be true
OR → any condition can be true

AGGREGATE FUNCTIONS

Prompt

Objective: Perform statistical analysis using aggregate functions.

```python
cursor.execute("""
SELECT COUNT(*), AVG(marks), MAX(marks), MIN(marks)
FROM students
""")

print(cursor.fetchall())
```

Key Learning
Used for analytics and reporting
Very important in interviews

GROUP BY

Prompt

Objective: Group students by department and calculate average marks.

```python
cursor.execute("""
SELECT department, AVG(marks)
FROM students
GROUP BY department
""")

print(cursor.fetchall())
```

key Learning
Groups similar data
Used with aggregate functions

HAVING

Prompt

Objective: Filter grouped data based on average marks.

```python
cursor.execute("""
SELECT department, AVG(marks)
FROM students
GROUP BY department
HAVING AVG(marks) > 85
""")

print(cursor.fetchall())
```

Key Learning
Filters grouped results
Used after GROUP BY

UPDATE

Prompt

Objective: Update marks of a specific student using their ID.

```python
cursor.execute("""
UPDATE students
SET marks = 92
WHERE student_id = 102
""")

conn.commit()
print("Updated successfully")
```

Key Learning
Modifies existing records
Always use WHERE

DELETE

Prompt

Objective: Delete a student record based on student ID.

```python
cursor.execute("""
DELETE FROM students
WHERE student_id = 104
""")

conn.commit()
print("Deleted successfully")
```

Key Learning
Removes data permanently
WHERE prevents full deletion

SUBQUERY

Prompt

Objective: Find students whose marks are above average using subquery.

```python
cursor.execute("""
SELECT name, marks
FROM students
WHERE marks > (SELECT AVG(marks) FROM students)
""")

print(cursor.fetchall())
```

Key Learning
Query inside another query
Used for complex logic

TOP PERFORMER

Prompt

Objective: Find student(s) with highest marks.

```python
cursor.execute("""
SELECT * FROM students
WHERE marks = (SELECT MAX(marks) FROM students)
""")

print(cursor.fetchall())
```

Key Learning
Finds best performer
Common interview question
